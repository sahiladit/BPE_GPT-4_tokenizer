```python
para = "The quick brown fox jumps over the lazy dog. Programming in Python makes string manipulation straightforward, even when handling complex character sets. When working with UTF-16, remember that characters are typically represented by 2 or 4 bytes! 🚀 Let's test this text."

en = para.encode('utf-8')
tokens = list(map(int,en))

print(tokens)
```

```python
def countFreq(tokens):
    freq = {}
    for pair in zip(tokens, tokens[1:]):
        freq[pair] = freq.get(pair,0) + 1

    return freq

freq = countFreq(tokens)

print(freq)
```

```python
# reversing the key (pair) and value (freq) to order the pairs->freq in ascending order

print(sorted(((v,k) for (k,v) in freq.items()), reverse = True))
```

```python
# BPE algo -> traverse over the tokens list, if we find 'e ' (MCP) then swap it with Z

def appendNewInit(tokens, pair, newVal):
    newToken = []
    i = 0

    while i < len(tokens):
        if i < len(tokens) - 1 and (tokens[i], tokens[i + 1]) == pair:
            newToken.append(newVal)
            i += 2
        else:
            newToken.append(tokens[i])
            i += 1

    return newToken
```

```python
# get top pair 
top = max(freq, key = freq.get)
top
```

```python
newToken = list(tokens)
newToken = appendNewInit(newToken, top, 256)
print(newToken)
print('New Length ', len(newToken))
```

```python
# now iterate this process 

vocab_size = 285
curr_size = vocab_size - 256
ids = list(tokens) # copy

merges = {} # (int,int) -> int (freq)

print("Original length:", len(tokens))

for i in range(vocab_size - 256):
    freq = countFreq(ids)

    if not freq:
        break

    top = max(freq, key=freq.get)
    idx = 256 + i

    print(
        f"Merging {top} -> {idx}, "
        f"frequency={freq[top]}, "
        f"length: {len(ids)} -> ",
        end=""
    )

    ids = appendNewInit(ids, top, idx)

    print(len(ids))

    merges[top] = idx

print("\nFinal length:", len(ids))
print("Tokens reduced by:", len(tokens) - len(ids))
```

**Decoding**

Given a sequence of integers in the range [0,vocab_size], convert it to plain text

```python
def decode(ids):
    # Reverse the merges:
    # 256 -> (110, 32)
    # 257 -> (32, 116)
    # etc.
    vocab = {v: k for k, v in merges.items()}

    def decode_token(token):
        # Original UTF-8 byte
        if token < 256:
            return [token]

        # Learned BPE token
        pair = vocab[token]

        return decode_token(pair[0]) + decode_token(pair[1])

    decoded_bytes = []

    for token in ids:
        decoded_bytes.extend(decode_token(token))

    return bytes(decoded_bytes).decode("utf-8")

s = decode(ids)
```

```python
def encode(text):
    # 1. Convert text to UTF-8 bytes
    ids = list(text.encode("utf-8"))

    # 2. Apply learned merges in training order
    while len(ids) >= 2:

        # Find the highest-priority available merge
        best_pair = None
        best_id = float("inf")

        for pair, token_id in merges.items():
            if token_id < best_id:
                # Check whether pair exists in current sequence
                if pair in zip(ids, ids[1:]):
                    best_pair = pair
                    best_id = token_id

        # No applicable merges
        if best_pair is None:
            break

        # 3. Merge it
        ids = appendNewInit(
            ids,
            best_pair,
            merges[best_pair]
        )

    return ids
```
